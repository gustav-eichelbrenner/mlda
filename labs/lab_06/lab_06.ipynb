# Генерируем данные

```python
# Create the dataset
X = np.linspace(0, 6, 200)[:, np.newaxis]
y = np.sin(X).ravel() + np.sin(6 * X).ravel() + np.random.RandomState(1).normal(0, 0.1, X.shape[0]) + 3
```

# Train / test split

```python
from sklearn.model_selection import train_test_split

# Split data into train and test samples
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.5, random_state=42)
```

# Обучаем регрессоры

Будем использовать три модели:

- kNN
- Дерево решений
- Линейная регрессия

Мы будем использовать scikit-learn реализацию этих регрессоров

Давайте просто импортируем их и обучим.

```python
# Import kNN regressor
from sklearn.neighbors import KNeighborsRegressor

# Create object of the regressor's class
knn_reg = KNeighborsRegressor(n_neighbors=10)

# Fit the regressor
knn_reg.fit(X_train, y_train)
```

```python
# Import Decision Tree regressor
from sklearn.tree import DecisionTreeRegressor

# Create object of the regressor's class
dt_reg = DecisionTreeRegressor(criterion='squared_error', splitter='best', max_depth=4, 
                            min_samples_split=2, min_samples_leaf=1)

# Fit the regressor
dt_reg.fit(X_train, y_train)
```

```python
# Import Linear Regression regressor
from sklearn.linear_model import LinearRegression

# Create object of the regressor's class
linreg = LinearRegression()

# Fit the regressor
linreg.fit(X_train, y_train)
```

# Прогнозы

```python
# kNN
y_test_knn_reg = knn_reg.predict(X_test)

# DT
y_test_dt_reg = dt_reg.predict(X_test)

# LinReg
y_test_linreg = linreg.predict(X_test)
```

```python
plt.figure(figsize=(15, 6))
plt.scatter(X_test[:, 0], y_test, color='0', label='Truth')

sortd_inds = np.argsort(X_test[:, 0])
plt.plot(X_test[sortd_inds, 0], y_test_knn_reg[sortd_inds], linewidth=3, color='b', label='kNN')
plt.plot(X_test[sortd_inds, 0], y_test_dt_reg[sortd_inds], linewidth=3, color='r', label='DT')
plt.plot(X_test[sortd_inds, 0], y_test_linreg[sortd_inds], linewidth=3, color='g', label='LinReg')

plt.xlabel('X', size=18)
plt.ylabel('y', size=18)
plt.legend(loc='best', fontsize=14)
plt.grid()
plt.show()
```

# Метрики качества

**1. (R)MSE ((Root) Mean Squared Error)**

$$ L(\hat{y}, y) = \frac{1}{N}\sum\limits_{n=1}^N (y_n - \hat{y}_n)^2$$

**2. MAE (Mean Absolute Error)**

$$ L(\hat{y}, y) = \frac{1}{N}\sum\limits_{n=1}^N |y_n - \hat{y}_n|$$

**3. RSE (Relative Squared Error)**

$$ L(\hat{y}, y) = \sqrt\frac{\sum\limits_{n=1}^N (y_n - \hat{y}_n)^2}{\sum\limits_{n=1}^N (y_n - \bar{y})^2}$$

**4. RAE (Relative Absolute Error)**

$$ L(\hat{y}, y) = \frac{\sum\limits_{n=1}^N |y_n - \hat{y}_n|}{\sum\limits_{n=1}^N |y_n - \bar{y}|}$$

**5. MAPE (Mean Absolute Persentage Error)**

$$ L(\hat{y}, y) = \frac{100}{N} \sum\limits_{n=1}^N\left|\frac{ y_n - \hat{y}_n}{y_n}\right|$$


**6. RMSLE (Root Mean Squared Logarithmic Error)**

$$ L(\hat{y}, y) = \sqrt{\frac{1}{N}\sum\limits_{n=1}^N(\log(y_n + 1) - \log(\hat{y}_n + 1))^2}$$

```python
from sklearn.metrics import mean_squared_error, mean_absolute_error, mean_squared_log_error

def regression_quality_metrics_report(y_true, y_pred):
    y_true = np.asarray(y_true)
    y_pred = np.asarray(y_pred)

    rmse = np.sqrt(mean_squared_error(y_true, y_pred))
    mae = mean_absolute_error(y_true, y_pred)
    rse = np.sqrt(np.sum((y_true - y_pred) ** 2) / np.sum((y_true - np.mean(y_true)) ** 2))
    rae = np.sum(np.abs(y_true - y_pred)) / np.sum(np.abs(y_true - np.mean(y_true)))
    mape = 100 * np.mean(np.abs((y_true - y_pred) / y_true))
    rmsle = np.sqrt(mean_squared_log_error(y_true, y_pred))

    return [rmse, mae, rse, rae, mape, rmsle]
```

```python
metrics_report = pd.DataFrame(columns=['RMSE', 'MAE', 'RSE', 'RAE', 'MAPE', 'RMSLE'])

metrics_report.loc['kNN', :] = regression_quality_metrics_report(y_test, y_test_knn_reg)
metrics_report.loc['DT', :] = regression_quality_metrics_report(y_test, y_test_dt_reg)
metrics_report.loc['LinReg', :] = regression_quality_metrics_report(y_test, y_test_linreg)

metrics_report
```

Вопросы:

1. Какой регрессор лучше?

kNN и дерево решений (DT) показывают близкое качество, а линейная регрессия заметно хуже обоих.

kNN чуть лучше по метрикам, основанным на квадрате ошибки: RMSE 0.4197 против 0.4266 и RSE 0.4079 против 0.4147. Эти метрики сильнее штрафуют крупные ошибки.
DT лучше по метрикам, основанным на модуле ошибки, и по относительным: MAE 0.326 против 0.356, RAE 0.386 против 0.421, MAPE 11.9% против 15.3%, RMSLE 0.112 против 0.121. Значит, у дерева в среднем ошибка меньше, но встречаются отдельные крупные промахи.
LinReg проигрывает по всем метрикам: RMSE 0.81, MAPE 29.6%. Из RSE ≈ 0.785 следует, что она объясняет только около 38% дисперсии (1 − RSE² ≈ 0.38). У kNN и DT этот показатель около 83%.

Однозначного победителя между kNN и DT нет. Если важна средняя ошибка, лучше DT. Если нужно избегать больших выбросов ошибки, лучше kNN. В целом на этих данных я бы выбрал DT, потому что он выигрывает по большему числу метрик.

3. Как объяснить поведение регрессоров?

LinReg строит прямую, а истинная зависимость y = sin(x) + sin(6x) + 3 колеблется. Прямая может уловить только общий уровень и слабый тренд. Это сильное недообучение (высокое смещение).
kNN при k = 10 усредняет ответы по окну примерно в 0.6 единицы по оси X. Период быстрой компоненты sin(6x) около 1.05, поэтому усреднение сглаживает её пики и впадины. Медленная компонента sin(x) восстанавливается хорошо.
DT с max_depth = 4 имеет не более 16 листьев и даёт ступенчатую аппроксимацию. Ступеньки грубо повторяют колебания, поэтому ошибка в среднем небольшая, но на пиках встречаются заметные отклонения.

4. Что будет при изменении гиперпараметров?

Ниже общее поведение моделей, на ваших данных я это не проверял.

kNN:
Меньше k (например, 2–3): модель точнее следует за sin(6x), ошибка на тесте снижается. При k = 1 она начинает подстраиваться под шум и переобучается.
Больше k (30 и выше): кривая сильнее сглаживается, быстрая компонента пропадает, качество падает и приближается к уровню линейной регрессии.
DT:
Больше max_depth (6–8): листьев больше, форма кривой воспроизводится лучше, метрики улучшаются. При очень большой глубине дерево начинает запоминать шум.
Меньше max_depth (1–2): модель слишком грубая, качество резко падает.
min_samples_leaf больше 1 ограничивает переобучение при большой глубине.
LinReg почти не имеет гиперпараметров. Улучшить её можно только изменением признаков.
