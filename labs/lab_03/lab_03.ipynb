Распределите студентов в соответствии с первым приоритетом. После распределения студентов в соответствии с первым приоритетом добавьте в свой датафрейм аналогичные признаки и запустите проверку:

```python
def make_capacity(df, season, special):
    """Вместимость курсов: по умолчанию 30, special — исключения"""
    cols = [f'{season}_1', f'{season}_2', f'{season}_3']
    courses = pd.unique(df[cols].to_numpy().ravel())
    capacity = pd.Series(30.0, index=courses)
    for course, seats in special.items():
        capacity[course] = seats
    return capacity


def distribute(df, season, capacity, n_waves=3, banned_ids=None, banned_course=None):
    """Возвращает таблицу «студент — курс» с признаком admitted после n_waves волн"""
    num = f'{season}_course_num'
    req = df.melt(id_vars=['id', 'percentile', num],
                  value_vars=[f'{season}_1', f'{season}_2', f'{season}_3'],
                  var_name='slot', value_name='course')

    # запрещённый курс для части студентов (нужен для весны), повторы курса у одного студента
    if banned_ids is not None:
        req = req[~(req['id'].isin(banned_ids) & (req['course'] == banned_course))]
    req = req.sort_values(['id', 'slot']).drop_duplicates(['id', 'course'])

    # волна: при одном курсе 1, 2, 3; при двух курсах fall_1 и fall_2 — первая волна, fall_3 — вторая
    priority = req['slot'].str[-1].astype(int)
    req['wave'] = np.where(req[num] == 1, priority, (priority + 1) // 2)

    req['admitted'] = False
    req = req.sort_values('percentile', ascending=True, kind='stable')   # лучшие, с наименьшим перцентилем, первыми   # лучшие по перцентили — первыми

    for wave in range(1, n_waves + 1):
        got = req.groupby('id')['admitted'].transform('sum')                # сколько курсов у студента уже есть
        taken = req.groupby('course')['admitted'].transform('sum')          # сколько мест уже занято
        seats_left = req['course'].map(capacity) - taken

        candidates = req[(req['wave'] == wave) & (got < req[num])]          # кому ещё нужен курс
        place = candidates.groupby('course').cumcount()                     # место в очереди на курс: 0, 1, 2...
        req.loc[candidates.index, 'admitted'] = place < seats_left[candidates.index]
    return req

check_df = pd.read_csv('./data/lab03_train_check.csv')
df['percentile'] = df['id'].map(check_df.set_index('id')['percentile'])

fall_capacity = make_capacity(df, 'fall', {'Statistical Learning Theory': 60,
                                           'Высокопроизводительные вычисления': 60,
                                           'Анализ неструктурированных данных': np.inf})

wave1 = distribute(df, 'fall', fall_capacity, n_waves=1)
enrolled = wave1[wave1['admitted']]

best = enrolled.groupby('course')['percentile'].transform('min')
worst = enrolled.groupby('course')['percentile'].transform('max')

df['is_first_place'] = df['id'].isin(enrolled.loc[enrolled['percentile'] == best, 'id']).map({True: True})
df['is_last_place'] = df['id'].isin(enrolled.loc[enrolled['percentile'] == worst, 'id']).map({True: True})

score += 2
print(f"Выполнено.\nscore = {score}")
```

Проведите все три волны отбора студентов на курсы по выбору. Результат сохраните в файл res_fall.csv.

```python
def make_result(df, req, season):
    """Таблица ID, course1, course2 в нужном формате"""
    num = f'{season}_course_num'
    enrolled = req[req['admitted']].copy()
    enrolled['n'] = enrolled.groupby('id').cumcount() + 1

    res = (enrolled.pivot(index='id', columns='n', values='course')
                   .reindex(index=df['id'], columns=[1, 2])
                   .rename(columns={1: 'course1', 2: 'course2'}))

    k = df.set_index('id')[num]
    res['course1'] = res['course1'].fillna('???')
    res['course2'] = res['course2'].fillna('???').mask(k == 1, '-')
    return res.reset_index().rename(columns={'id': 'ID'})


req_fall = distribute(df, 'fall', fall_capacity)
res_fall = make_result(df, req_fall, 'fall')
res_fall.to_csv('res_fall.csv', index=None)
res_fall.head()

score += 1
print(f"Выполнено.\nscore = {score}")
```
